```python
import matplotlib.pyplot as plt
import numpy as np
import torch
```

```python
start = 0.0001
end = 0.02
step = 1000
```

```python
plt.plot([*range(step)], torch.linspace(start,end,step))
plt.xlabel('timestep')
plt.ylabel('beta')
plt.show()
```

```python
x = torch.linspace(0, step, step+1)
alphas_cumprod = torch.cos(((x / step) + 0.008) / (1 + 0.008) * torch.pi * 0.5) ** 2
alphas_cumprod = alphas_cumprod / alphas_cumprod[0]
betas = 1 - (alphas_cumprod[1:] / alphas_cumprod[:-1])
plt.plot([*range(step)], torch.clip(betas, 0.0001, 0.9999))
plt.xlabel('timestep')
plt.ylabel('beta')
plt.show()
```
